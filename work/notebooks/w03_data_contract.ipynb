## Leakage Experiment

I deliberately added **trend_direction** to the feature set.

This column is derived from the outcome represented by **trend_pct**, so it leaks future information into the model.

A model using this feature would appear unrealistically accurate.

After removing **trend_direction**, the remaining feature set represents information that would genuinely be available at the decision time.

```python
from sklearn.linear_model import LinearRegression

# Honest features
X = df[
    [
        "impressions_90d",
        "clicks_90d",
        "sessions_90d",
        "avg_position",
        "engagement_rate"
    ]
]

y = df["trend_pct"]

print("Honest feature matrix shape:", X.shape)

# Deliberately leaked feature
X_leak = X.copy()
X_leak["trend_direction"] = pd.factorize(df["trend_direction"])[0]

print("Feature matrix with leakage:", X_leak.shape)

# Remove leakage
X_fixed = X_leak.drop(columns=["trend_direction"])

print("Leakage removed.")
print("Final feature matrix shape:", X_fixed.shape)
```
