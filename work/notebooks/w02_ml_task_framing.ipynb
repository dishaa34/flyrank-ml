## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# Display correlations between important numeric features

cols = [
    "impressions_90d",
    "clicks_90d",
    "sessions_90d",
    "ai_sessions_90d",
    "ai_traffic_pct"
]

df[cols].corr()
```

## Why ML Beats a Fixed Rule

A simple rule such as "refresh every page with more than 10,000 impressions" ignores many other important factors.

The opportunity to refresh content depends on multiple features working together, including impressions, clicks, sessions, and AI traffic.

Machine learning can learn patterns across all these variables and rank pages more effectively than a single threshold-based rule.

The output supports content teams by helping them prioritize which pages should be refreshed first.
